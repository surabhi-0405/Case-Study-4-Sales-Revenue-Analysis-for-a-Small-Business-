```python
q25 = customer_metrics['Total_Revenue'].quantile(0.25)
q50 = customer_metrics['Total_Revenue'].quantile(0.50)
q75 = customer_metrics['Total_Revenue'].quantile(0.75)

customer_metrics['Revenue_Segment'] = pd.cut(
    customer_metrics['Total_Revenue'],
    bins=[-float('inf'), q25, q50, q75, float('inf')],
    labels=['Low Value', 'Regular Value', 'High Value', 'Very High Value'],
    include_lowest=True
)

customer_metrics['Revenue_Segment'].value_counts().sort_index()
```

```python
frequency_median = customer_metrics['Transaction_Frequency'].median()

customer_metrics['Customer_Segment'] = 'Occasional'

customer_metrics.loc[
    (customer_metrics['Transaction_Frequency'] > frequency_median) &
    (customer_metrics['Total_Revenue'] > q50),
    'Customer_Segment'
] = 'High Value Frequent'

customer_metrics.loc[
    (customer_metrics['Transaction_Frequency'] <= frequency_median) &
    (customer_metrics['Total_Revenue'] > q50),
    'Customer_Segment'
] = 'High Value'

customer_metrics.loc[
    (customer_metrics['Transaction_Frequency'] > frequency_median) &
    (customer_metrics['Total_Revenue'] <= q50),
    'Customer_Segment'
] = 'Frequent'

customer_metrics['Customer_Segment'].value_counts()
```

## **Future Sales Forecasting And Predictive CLV**

```python
import pandas as pd

sales_df = pd.read_csv('/content/TrendMart_Sales_Cleaned.csv')

sales_df.head()
```

```python
# Make sure Date is in datetime format
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# Calculate monthly revenue
monthly_sales = (
    sales_df
    .groupby(sales_df['Date'].dt.to_period('M'))['Total_Amount']
    .sum()
    .reset_index()
)

monthly_sales['Date'] = monthly_sales['Date'].dt.to_timestamp()

monthly_sales
```

```python
# Prepare monthly sales data for forecasting

forecast_df = monthly_sales.copy()

forecast_df['Month_Index'] = range(1, len(forecast_df) + 1)

forecast_df
```

```python
from sklearn.linear_model import LinearRegression

# Prepare training data
X = forecast_df[['Month_Index']]
y = forecast_df['Total_Amount']

# Train the model
model = LinearRegression()
model.fit(X, y)

print("Model trained successfully.")
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Forecast the next 3 months

future_months = pd.DataFrame({
    'Month_Index': [5, 6, 7]
})

future_months['Forecasted_Revenue'] = model.predict(
    future_months[['Month_Index']]
)

future_months
```

```python
# Add forecast dates

last_date = monthly_sales['Date'].max()

future_months['Forecast_Date'] = [
    last_date + pd.DateOffset(months=i)
    for i in range(1, 4)
]

future_months
```

```python
# Prepare actual sales data
actual = monthly_sales.rename(
    columns={'Total_Amount': 'Revenue'}
)[['Date', 'Revenue']]

actual['Type'] = 'Actual'

# Prepare forecast data
forecast = future_months.rename(
    columns={
        'Forecast_Date': 'Date',
        'Forecasted_Revenue': 'Revenue'
    }
)[['Date', 'Revenue']]

forecast['Type'] = 'Forecast'

# Combine actual and forecast
sales_forecast = pd.concat(
    [actual, forecast],
    ignore_index=True
)

sales_forecast
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

# Predictions for historical months
historical_predictions = model.predict(
    forecast_df[['Month_Index']]
)

# Model evaluation
r2 = r2_score(
    forecast_df['Total_Amount'],
    historical_predictions
)

mae = mean_absolute_error(
    forecast_df['Total_Amount'],
    historical_predictions
)

print("R² Score:", round(r2, 4))
print("Mean Absolute Error:", round(mae, 2))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

# Actual revenue
plt.plot(
    sales_forecast[sales_forecast['Type'] == 'Actual']['Date'],
    sales_forecast[sales_forecast['Type'] == 'Actual']['Revenue'],
    marker='o',
    label='Actual Revenue'
)

# Forecast revenue
plt.plot(
    sales_forecast[sales_forecast['Type'] == 'Forecast']['Date'],
    sales_forecast[sales_forecast['Type'] == 'Forecast']['Revenue'],
    marker='o',
    linestyle='--',
    label='Forecast Revenue'
)

plt.title('TrendMart Sales Revenue Forecast')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

# Ensure date format
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# ---------------------------------------------------------
# 1. Define training and future periods
# ---------------------------------------------------------
cutoff_date = pd.Timestamp('2024-03-01')

train_data = sales_df[sales_df['Date'] < cutoff_date]
future_data = sales_df[sales_df['Date'] >= cutoff_date]

# ---------------------------------------------------------
# 2. Create customer features from Jan-Feb
# ---------------------------------------------------------
customer_features = (
    train_data
    .groupby('Customer_ID')
    .agg(
        Historical_Revenue=('Total_Amount', 'sum'),
        Transaction_Frequency=('Transaction_ID', 'nunique'),
        Average_Order_Value=('Total_Amount', 'mean'),
        Total_Quantity=('Quantity', 'sum')
    )
    .reset_index()
)

# ---------------------------------------------------------
# 3. Add customers with no Jan-Feb transactions
# ---------------------------------------------------------
all_customers = sales_df['Customer_ID'].unique()

customer_features = (
    pd.DataFrame({'Customer_ID': all_customers})
    .merge(customer_features, on='Customer_ID', how='left')
    .fillna(0)
)

# ---------------------------------------------------------
# 4. Target = March-April revenue
# ---------------------------------------------------------
future_revenue = (
    future_data
    .groupby('Customer_ID')['Total_Amount']
    .sum()
    .reset_index()
    .rename(columns={'Total_Amount': 'Future_Revenue'})
)

customer_features = customer_features.merge(
    future_revenue,
    on='Customer_ID',
    how='left'
)

customer_features['Future_Revenue'] = (
    customer_features['Future_Revenue'].fillna(0)
)

# ---------------------------------------------------------
# 5. Train predictive model
# ---------------------------------------------------------
features = [
    'Historical_Revenue',
    'Transaction_Frequency',
    'Average_Order_Value',
    'Total_Quantity'
]

X = customer_features[features]
y = customer_features['Future_Revenue']

clv_model = LinearRegression()
clv_model.fit(X, y)

# ---------------------------------------------------------
# 6. Evaluate prediction
# ---------------------------------------------------------
predicted_revenue = clv_model.predict(X)

r2 = r2_score(y, predicted_revenue)
mae = mean_absolute_error(y, predicted_revenue)

print("Predictive CLV model trained successfully.")
print("R² Score:", round(r2, 4))
print("Mean Absolute Error:", round(mae, 2))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

# Features and target
X = customer_features[features]
y = customer_features['Future_Revenue']

# Split customers into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train model
clv_model = LinearRegression()
clv_model.fit(X_train, y_train)

# Predict on unseen customers
test_predictions = clv_model.predict(X_test)

# Evaluation
test_r2 = r2_score(y_test, test_predictions)
test_mae = mean_absolute_error(y_test, test_predictions)

print("Predictive CLV test evaluation")
print("R² Score:", round(test_r2, 4))
print("Mean Absolute Error:", round(test_mae, 2))
```

```python
# Train final model using all available customer history
final_clv_model = LinearRegression()
final_clv_model.fit(
    customer_features[features],
    customer_features['Future_Revenue']
)

# Predict future revenue for all customers
customer_features['Predicted_Future_Revenue'] = (
    final_clv_model.predict(customer_features[features])
)

# Prevent negative predicted values
customer_features['Predicted_Future_Revenue'] = (
    customer_features['Predicted_Future_Revenue'].clip(lower=0)
)

# Create a simple predicted-value segment
customer_features['Predicted_Value_Segment'] = pd.cut(
    customer_features['Predicted_Future_Revenue'],
    bins=[-1, 500, 1500, float('inf')],
    labels=['Low', 'Medium', 'High']
)

# Display highest predicted-value customers
customer_features[
    [
        'Customer_ID',
        'Historical_Revenue',
        'Transaction_Frequency',
        'Average_Order_Value',
        'Total_Quantity',
        'Predicted_Future_Revenue',
        'Predicted_Value_Segment'
    ]
].sort_values(
    'Predicted_Future_Revenue',
    ascending=False
).head(10)
```

```python
predictive_clv = customer_features[
    [
        'Customer_ID',
        'Historical_Revenue',
        'Transaction_Frequency',
        'Average_Order_Value',
        'Total_Quantity',
        'Future_Revenue',
        'Predicted_Future_Revenue',
        'Predicted_Value_Segment'
    ]
].copy()

predictive_clv.to_csv(
    '/content/TrendMart_Predictive_CLV.csv',
    index=False
)

print("Predictive CLV dataset saved successfully.")
print("Rows:", len(predictive_clv))
print("Columns:", len(predictive_clv.columns))
```
